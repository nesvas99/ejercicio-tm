```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# Datos de tiempo de viaje entre estaciones (en minutos)
tiempos = np.array([[51],[47],[54],[49],[52],[50],[47],[53],[55],[50],[53],[46],[55],[49],[51],[52],[50],[48],[52],[49],[53],[50],[47],[51],[49],[52],[50],[53],[47],[50],[49],[51],[52],[50],[48],[52],[49],[53],[50],[47],[51],[54],[47],[50],[49],[51],[52],[50],[48],[52],[49],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50],[47],[51],[49],[52],[50],[53],[50]
])
# Tiempos de viaje (en minutos)
viajes = np.array([48,45,53,49,51,49,47,52,54,50,53,46,55,48,50,51,49,47,52,49,52,50,47,50,48,52,49,53,46,50,48,50,52,49,47,51,49,52,50,47,50,53,46,50,48,50,52,49,47,51,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50,47,50,48,52,49,52,50
])
# Dividir los datos en conjuntos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(tiempos, viajes, test_size=0.2, random_state=150)

# Crear el modelo de regresión lineal
modelo = LinearRegression()

# Entrenar el modelo
modelo.fit(X_train, y_train)

# Hacer predicciones
y_pred = modelo.predict(X_test)

# Evaluar el modelo
mse = mean_squared_error(y_test, y_pred)
print(f"Error cuadrático medio: {mse}")

# Visualizar los resultados
plt.scatter(tiempos, viajes, color='blue', label='Datos reales')
plt.plot(X_test, y_pred, color='red', label='Predicciones')
plt.xlabel('Tiempo de viaje (minutos)')
plt.ylabel('Tiempo de viaje predicho (minutos)')
plt.title('Regresión Lineal: Tiempo de viaje entre estaciones de TransMilenio')
plt.legend()
plt.show()
```
